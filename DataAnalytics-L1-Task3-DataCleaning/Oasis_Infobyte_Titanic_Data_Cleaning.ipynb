```python
print("Missing values after Step 3:")
print(df.isnull().sum())
```

```python
# ================================
# STEP 4: COMPLETE DATA CLEANING
# ================================

# 1. Check and remove duplicate rows
duplicates_before = df.duplicated().sum()
df = df.drop_duplicates()
duplicates_after = df.duplicated().sum()

print("Duplicate rows before cleaning:", duplicates_before)
print("Duplicate rows after cleaning:", duplicates_after)


# 2. Standardize categorical columns
# Remove unnecessary spaces and standardize text
for column in ['Sex', 'Embarked']:
    df[column] = df[column].astype(str).str.strip()

df['Sex'] = df['Sex'].str.lower()
df['Embarked'] = df['Embarked'].str.upper()

print("\nUnique Sex values:", df['Sex'].unique())
print("Unique Embarked values:", df['Embarked'].unique())


# 3. Correct data types
df['PassengerId'] = df['PassengerId'].astype(int)
df['Survived'] = df['Survived'].astype(int)
df['Pclass'] = df['Pclass'].astype(int)
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df['SibSp'] = pd.to_numeric(df['SibSp'], errors='coerce')
df['Parch'] = pd.to_numeric(df['Parch'], errors='coerce')
df['Fare'] = pd.to_numeric(df['Fare'], errors='coerce')


# 4. Check logical ranges
print("\nRange checks:")
print("Age range:", df['Age'].min(), "to", df['Age'].max())
print("Fare range:", df['Fare'].min(), "to", df['Fare'].max())
print("Pclass values:", sorted(df['Pclass'].unique()))
print("Survived values:", sorted(df['Survived'].unique()))


# 5. Outlier detection using IQR
numeric_columns = ['Age', 'Fare', 'SibSp', 'Parch']

outlier_report = {}

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = ((df[column] < lower_bound) |
                (df[column] > upper_bound)).sum()

    outlier_report[column] = outliers

print("\nOutlier count:")
print(outlier_report)


# 6. Cap extreme outliers using IQR
# This keeps the records instead of deleting them.

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df[column] = df[column].clip(lower=lower_bound, upper=upper_bound)

print("\nOutliers have been capped using the IQR method.")

print("\nCleaning completed successfully!")
```

```python
# ==========================================
# STEP 5: BEFORE vs AFTER QUALITY REPORT
# ==========================================

quality_report = pd.DataFrame({
    'Metric': [
        'Number of Rows',
        'Number of Columns',
        'Missing Values',
        'Duplicate Rows'
    ],
    'After Cleaning': [
        len(df),
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum()
    ]
})

print("DATA QUALITY REPORT AFTER CLEANING")
display(quality_report)

print("\nMissing values by column:")
display(df.isnull().sum())

print("\nData types after cleaning:")
display(df.dtypes)
```

```python
# ==========================================
# STEP 6: FINAL CLEANED DATASET
# ==========================================

print("Final dataset shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())

print("\nFinal missing values:")
print(df.isnull().sum())

print("\nFinal duplicate count:", df.duplicated().sum())

# Save cleaned dataset
df.to_csv('titanic_cleaned.csv', index=False)

print("\nCleaned dataset saved successfully as: titanic_cleaned.csv")
```

# Titanic Dataset Cleaning and Preprocessing

## Project Overview

This project focuses on cleaning and preprocessing the Titanic dataset to improve its data quality and prepare it for further analysis.

The dataset was inspected for missing values, duplicate records, inconsistent formatting, data type issues, and numerical outliers.

## Objectives

- Identify data quality issues.
- Handle missing values appropriately.
- Check and remove duplicate records.
- Standardize inconsistent categorical values.
- Detect and handle numerical outliers using the IQR method.
- Verify and correct data types.
- Create a clean dataset ready for analysis.

## Dataset Cleaning Decisions

### Missing Values
- Missing values in `Age` were filled using the median.
- Missing values in `Embarked` were filled using the mode.
- Missing values in `Cabin` were replaced with `Unknown`.

### Duplicate Records
Duplicate rows were checked and removed if present.

### Data Standardization
The `Sex` column was standardized to lowercase, while the `Embarked` column was standardized to uppercase.

### Outliers
Outliers in numerical columns were detected using the Interquartile Range (IQR) method. Extreme values were capped instead of removing records to preserve valid passenger information.

## Conclusion

The Titanic dataset was successfully cleaned and prepared for further analysis. Missing values, duplicates, formatting inconsistencies, data types, and numerical outliers were addressed.

The final cleaned dataset is suitable for exploratory analysis and machine learning applications.
